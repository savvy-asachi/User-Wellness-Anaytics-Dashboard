```python
user_summary=(
    df.groupby("Id").agg({
  "TotalSteps":["mean","max","min"],
   "Calories":"mean"})    ) 



#improving how table looks making it proper

user_summary.columns=[
    "AvgSteps",
    "MaxSteps",
    "MinSteps",
    "AvgCalories"
]
user_summary.head()
```

```python
#function for activity level
def activity_level(steps):
    if steps<5000 :
        return "low"
    elif steps<10000 :
        return "Moderate"
    else:
        return "high"


        
            
```

```python
user_summary["ActivityLevel"]=user_summary["AvgSteps"].apply(activity_level)
user_summary.head()
```

```python
user_summary["ActivityLevel"].value_counts()
user_summary["AvgCalories"].mean()
```

```python
user_summary.sort_values("AvgSteps").head()
user_summary.sort_values("AvgCalories",ascending=False).head()

user_summary["StepsScore"]=(user_summary["AvgSteps"]/ user_summary["AvgSteps"] / user_summary["AvgSteps"].max())*100
```

```python
user_summary["CaloriesScore"]=(user_summary["AvgCalories"] /
                               user_summary["AvgCalories"].max())*100
```

```python
user_summary["WellnessScore"]=(user_summary["StepsScore"] * 0.6 + user_summary["CaloriesScore"] * 0.4 )
```

```python
user_summary.sort_values("WellnessScore",ascending=False).head()
```

```python
def risk_level(score):
    if score>= 70:
        return "Low risk"
    elif score>= 40:
        return "medium Risk"
    else: 
        return "high level"
        
user_summary["RiskLevel"]=user_summary["WellnessScore"].apply(risk_level)
```

```python
user_summary[["WellnessScore","RiskLevel"]].head()
```

```python
def user_segment(score):
    if score>=35:
        return "Highly Active"
    elif score>=30:
        return "Active"
    elif score>=25:
        return "Moderately Active"
    else:
        return "needs improvement"
        
user_summary["UserSegment"]=user_summary["WellnessScore"].apply(user_segment)
```

```python
user_summary[["WellnessScore","UserSegment"]].head()
```

```python
user_summary["UserSegment"].value_counts()
```

```python
kpis={
    "Total Users":
    user_summary.index.nunique(),
    "Average Steps":
    round(user_summary["AvgSteps"].mean(),2),
    "Average Calories":
    round(user_summary["AvgCalories"].mean(),2),
    "Average Wellness Score":
    round(user_summary["WellnessScore"].mean(),2)
}
kpis
    
```

```python
top_users=user_summary.sort_values(by="WellnessScore", ascending=False).head(10)
top_users                                    
                                    
```

```python
user_summary.to_csv("user_summary.csv",index=False)
top_users.to_csv("top10_users.csv",index=False)
```
